
**Block 3 — GPT-2**

````python
generator = pipeline("text-generation", model="gpt2")

result = generator(
    "You suck, but in a strangely funny way, my babe. Let me explain:",

    max_new_tokens=200,        # reduced (important)
    do_sample=True,
    temperature=0.7,           # controls randomness
    top_k=50,                  # avoids nonsense
    top_p=0.9,                 # nucleus sampling
    repetition_penalty=1.2,    # reduces loops
    no_repeat_ngram_size=3,    # prevents repeating phrases
    truncation=True
)

print(result[0]['generated_text'])
# ```
# You will see GPT-2 continuing your sentence.

# **Add a text cell:**
# ```
# GPT-2 is decoder-only. It only sees left context — predicts the next
# token one at a time. Good for generation. Cannot do classification well.
````

**Block 4 — T5**


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("t5-small")

text = """The model was trained on 10,000 samples but achieved only 60% accuracy
on the test set due to severe overfitting. Regularization techniques like
dropout and weight decay were not applied during training."""

inputs = tokenizer("summarize: " + text, return_tensors="pt", truncation=True)

outputs = model.generate(
    **inputs,
    max_new_tokens=40,
    min_length=10
)

summary = tokenizer.decode(outputs[0], skip_special_tokens=True)
print("T5 output:", summary)
````

**Block 5 — BART Zero-shot**

````python
zs = pipeline("zero-shot-classification",
              model="facebook/bart-large-mnli")

result = zs(
    "The transformer architecture has revolutionized natural language processing.",
    candidate_labels=["machine learning", "sports", "finance", "politics"]
)

print("Top label:", result['labels'][0])
print("Score:", round(result['scores'][0], 3))
print("\nAll labels and scores:")
for label, score in zip(result['labels'], result['scores']):
    print(f"  {label}: {round(score, 3)}")
# ```
# You will see "machine learning" ranked #1.

# **Add a text cell:**
# ```
# BART is also encoder-decoder. Zero-shot means no fine-tuning needed —
# you define the labels yourself at runtime. Extremely useful in practice.
````
